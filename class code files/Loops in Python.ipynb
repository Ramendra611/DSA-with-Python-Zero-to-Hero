````python
## given a list of numbers, find the minimum of the numbers


nums = [4, 6, 8, 10]


# Method 1 : Using in built function
min_number = min(nums)
print(min_number)


# Method 2: Writing loop but still using in built function
min_number = nums[0]
for num in nums:
    min_number = min(min_number, num)
print(min_number)


# Method 3: Without using in-built function
min_number = nums[0]
for num in nums:
    if num < min_number:
        min_number = num
print(min_number)
````

````python
'''Given the list of marks in 3 subjects for students, create a mapping of student and total and 
average marks and the result. 
If the student fails in any subject, then overall he is fail. 


result_dict = {

    "Anubhav": {
                    marks:
                    total_marks : <>, 
                    average_marks : <>, 
                    maximum_marks : <>, 
                    minimum_marks : <>, 
                    result = <pass/fail>, 
                    fail_subjects: <>
                    }, 

    "Navneet":
}
'''

names_list = ['Anubhav', 'Navneet', 'Deviprasad']
marks_list = [[50, 60 , 90] , [30, 50, 80], [40, 100, 20]]
passing_marks = 40 
result_dict = {}
for ind, marks in enumerate(marks_list):
    # print(marks)
    total_marks = 0
    min_marks, max_marks = marks[0], marks[0]
    result = "pass" # flag
    for mark in marks:
        total_marks = total_marks + mark
        min_marks = min(min_marks, mark)
        max_marks = max(max_marks, mark)
        if mark < passing_marks:
            result = "fail"
        
    average_marks = total_marks / len(marks)

    # print(f"{total_marks= }, {min_marks = }, {max_marks = } {average_marks = }   ")

    result_dict[names_list[ind]] = {
        "marks": marks, 
        "total_marks": total_marks, 
        "average_marks" : average_marks, 
        "min_marks": min_marks, 
        "max_marks" : max_marks,  
        "result" : result
    }

    # print("--" *20)
print(result_dict)
````

````python
'''
Write a function that will give me important statistics about a given list of numbers
    - count of elements
    - min of the elements
    - max of the elements
    - number of odd and even elements in the list
    - sum of the list
    - average of the list
    - median of the list # what is the meaning of median?
    - 75%ile in the list
'''
def get_min(list_):
    if list_: # calculate only the list has atleast one element
        min_ = list_[0]
        for i in list_:
            min_ = min(min_, i)
        return min_

def get_max(list_):
    if list_: # calculate only the list has atleast one element
        max_ = list_[0]
        for i in list_:
            max_ = max(max_, i)
        return max_

def is_even(num):
    '''
    This function checks if a number is even and returns True else returns False
    '''
    # if num % 2 == 0:
    #     return True
    # else:
    #     return False
    
    # return True if num % 2 == 0 else False
    
    return num % 2 == 0
    

def count_of_odd_even(list_):
    '''
    This function retuns the count of odd and even numbers in a given list

    Returns:
        
    '''
    count_odd, count_even = 0, 0
    for i  in list_:
        if is_even(i):
            count_even += 1
        else:
            count_odd += 1
    # return count_even, count_odd
    return { "count_even": count_even, "count_odd": count_odd }
         
def get_stats(list_):
    count = get_length(list_)
    min_ = get_min(list_)
    max_  = get_max(list_)

    ### getting even and odd count
    result = count_of_odd_even(list_)
    count_of_odd = result["count_odd"]
    count_of_even = result["count_even"]

    sum_ = get_sum(list_)
    average = get_average(list_)

    # return count, min_, max_, count_odd, count_even, sum_, average
    return  {

                "count": count,
                "min": min_,
                "max": max_,
                "count_odd": count_of_odd,
                "count_even":count_of_even,
                "sum":sum_, 
                "average": average,
        
    }

````

````python
## Given a string, write a function to determine if the string is a palindrome
## reverse of "madam" is "madam" or reverse of "racecar" is "racecar"

def reverse_string(string):
    '''
    This function takes a string and reverses the string and return the reversed_string
    '''
    reversed_string = ""
    for char in string:
        reversed_string = char + reversed_string
    return reversed_string


def is_palindrome(string):
    '''
    This function checks if the given string is a palindrome. If yes then returns True else False
    '''
    ## if the string is equal to its reverse then the string is palindrome
    reverse_of_string = reverse_string(string)
    return True if string.lower() == reverse_of_string.lower() else False

is_palindrome("racecar")    
````
