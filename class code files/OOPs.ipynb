```python
class Hospital:
    def __init__(self, name, list_of_patients = None):
        self.name = name
        self.patient_list = list_of_patients if list_of_patients else []

    def admit_patient(self, patient, doctor):
        self.patient_list.append(patient)
        patient.attending_doctor = doctor
        doctor.patients.append(patient)

    def __str__(self):
        return f"{self.name}"
    
    def __repr__(self):
        return f"{self.name}"
        

class Patient:
    def __init__(self, name, complaints = None):
        self.name = name
        self.complaints = complaints if complaints else []
        self.attending_doctor = None  

    def __str__(self):
        return f"Patient = {self.name}"

    def __repr__(self):
        return f"Patient - {self.name}"

class Book:
    def __init__(self, name, writer):
        self.title = name
        self.author = writer


class Doctor:
    # special method/ magic method / constructor / init method
    def __init__(self, name: str, specialisation: str, hospital: str, patients = None):
        '''
        Init method is executed whenever a class is instantiated
        '''
        print("Creating a doctor with name :", name)
        self.name = name
        self.specialisation = specialisation
        self.hospital = hospital
        self.patients = patients if patients else []
        print(f"Doctor {self.name} is created!!!")

    # method
    def perform_surgery(self, patient = None):
        print(f"Dr. {self.name} is performing surgery!!")

    def reading_book(self, book =  None):
        print(f"Dr. {self.name} is reading {book.title} by the author {book.author}!!")

    def __str__(self):
        return f"( Dr.{self.name} | {self.specialisation} | {self.hospital} )"

    def __repr__(self):
        return f"[Dr.{self.name}|{self.specialisation}|{self.hospital}]"
    
```

## Types of methods
    - Instance Methods --> takes instance as a parameter
    - Class Methods --> take the class as a parameter
    - Static Methods --> takes neither class nor instance. it is like a normal function but it is bundles inside a class. ( Utility function. )

```python
class Employee:
    # class variables --> belongs to the class
    company_name = "codeverra"
    rating = "Meets Expectations"
    appraisal_percent = 0.1
    
    def __init__(self, name, deptt, salary, dob):
        ## instance attributes
        self.name = name
        self.deptt = deptt
        self.salary = salary
        self.dob  = dob 
        # self.age = self.calculate_age_from_dob(self.dob) # this is not the correct way to set the age
        self.email = f"{self.name}@{self.company_name}.com"

    def get_age(self):
        age = self.calculate_age_from_dob(self.dob)
        return age
    
    @staticmethod  ## using this decorator to indicate that this method is static method
    def calculate_age_from_dob(dob):
        yob = datetime.datetime.strptime(dob, "%Y-%m-%d").year
        current_year = datetime.datetime.now().year
        return current_year - yob

    def give_promotion(self):
        tmp = self.salary
        self.salary = self.salary * (1 + self.appraisal_percent)
        print(f"{self.name} is promoted! Salary increased from {tmp} to {self.salary}!")
        # self.salary = self.salary * (1 + Employee.appraisal_percent)

    def set_special_appraisal(self, special_appraisal):
        self.appraisal_percent = special_appraisal
    
    @classmethod #decorator (purpose is to add some extra functionality to the function)
    def update_appraisal_percentage(cls, updated_appraisal_percent):
        cls.appraisal_percent = updated_appraisal_percent

    def __str__(self):
        return f" {self.name}|{self.deptt}"

    ## alternate constructor
    @classmethod
    def create_employee_from_string(cls, input_string):
        name, deptt, salary, dob = input_string.split(",")
        return cls(name, deptt, salary, dob) ## returning a instance created from the input_string
        
    

```
